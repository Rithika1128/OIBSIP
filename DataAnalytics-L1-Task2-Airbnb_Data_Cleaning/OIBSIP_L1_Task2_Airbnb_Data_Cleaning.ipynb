```python
# Check whether missing reviews_per_month corresponds to listings with no reviews

missing_review_data = df[df['reviews_per_month'].isnull()]

print("Missing reviews_per_month:", len(missing_review_data))
print("Number of reviews for these listings:")
print(missing_review_data['number_of_reviews'].value_counts())
```

```python
# Check the relationship between the two review-related missing columns

print("Missing last_review:", df['last_review'].isnull().sum())
print("Missing reviews_per_month:", df['reviews_per_month'].isnull().sum())

print("\nRows where reviews_per_month is missing but number_of_reviews > 0:")
display(
    df[
        df['reviews_per_month'].isnull() &
        (df['number_of_reviews'] > 0)
    ][['id', 'number_of_reviews', 'last_review', 'reviews_per_month']]
)
```

```python
# Handle missing values

df['reviews_per_month'] = df['reviews_per_month'].fillna(0)

df['name'] = df['name'].fillna('Unknown')

df['host_name'] = df['host_name'].fillna('Unknown')

print("Missing values after handling:")
display(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
# Convert last_review to datetime

df['last_review'] = pd.to_datetime(df['last_review'], errors='coerce')

print("Data type of last_review:")
print(df['last_review'].dtype)

print("\nMissing values in last_review:")
print(df['last_review'].isnull().sum())
```

```python
# Check unique values in important categorical columns

print("Neighbourhood Groups:")
print(df['neighbourhood_group'].unique())

print("\nRoom Types:")
print(df['room_type'].unique())

print("\nNumber of unique neighbourhoods:", df['neighbourhood'].nunique())
```

```python
# Check for leading/trailing whitespace

for col in ['name', 'host_name', 'neighbourhood_group', 'neighbourhood', 'room_type']:
    whitespace_count = (
        df[col].dropna()
        .astype(str)
        .apply(lambda x: x != x.strip())
        .sum()
    )
    print(f"{col}: {whitespace_count} values with extra whitespace")
```

```python
# Find possible text encoding artifacts

for col in ['name', 'host_name']:
    mask = df[col].astype(str).str.contains(r'Ã|Â|â', regex=True, na=False)
    print(f"\n{col}: {mask.sum()} possible encoding issues")

    if mask.sum() > 0:
        display(df.loc[mask, ['id', col]].head(10))
```

```python
def fix_encoding(text):
    try:
        return text.encode('latin1').decode('utf-8')
    except:
        return text

df['name'] = df['name'].apply(fix_encoding)
df['host_name'] = df['host_name'].apply(fix_encoding)

print("Encoding cleanup completed.")
```

```python
# Detect outliers using the IQR method

numeric_cols = [
    'price',
    'minimum_nights',
    'number_of_reviews',
    'reviews_per_month',
    'calculated_host_listings_count',
    'availability_365'
]

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = ((df[col] < lower) | (df[col] > upper)).sum()

    print(f"{col}: {outliers} outliers")
```

## Outlier Detection and Treatment

Outliers were identified using the Interquartile Range (IQR) method.

Several numerical columns contained statistical outliers. However, these values were not automatically removed because unusual values may represent legitimate Airbnb listings or host behavior.

- `price`: High-priced listings may be legitimate.
- `minimum_nights`: Long-term stays may be valid.
- `number_of_reviews`: Popular listings can naturally have many reviews.
- `reviews_per_month`: High review activity can be legitimate.
- `calculated_host_listings_count`: Professional hosts may manage multiple listings.
- `availability_365`: No IQR outliers were detected.

Therefore, the statistical outliers were retained rather than capped or removed.

Listings with a price of 0 were handled separately and removed because a zero price is not meaningful for standard Airbnb price analysis.
